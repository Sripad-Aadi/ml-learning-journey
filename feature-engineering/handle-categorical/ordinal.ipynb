### Ordinal encoding: Encoding for features where order matters.

#### Ex. Education: School, College and Graduation => 0, 1 and 3 respectively

```python
df_cate = df.iloc[:,2:4]
df_cate.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(df_cate, df['purchased'],test_size=0.2,random_state=42)
```

```python
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
X_train_encoded=oe.fit_transform(X_train)
```
