#### fever needs to be simple imputer

To do all of these feature eng techniques again and again on training, testing and during production is quiet head ace so using all at once is done through column transformer

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import LabelEncoder
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
trans = ColumnTransformer([
    ('trans1',SimpleImputer(),['fever']),
    ('trans2',OneHotEncoder(sparse_output=False,drop='first',dtype=int),['gender','city']),
    ('trans3',OrdinalEncoder(categories=[['Mild','Strong']]),['cough'])
])
```

```python
X_train_trans = trans.fit_transform(X_train)
```
