```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset('tips')
tips.sample(5)
```

```python
flights=sns.load_dataset('flights')
flights.sample()
```

```python
iris=sns.load_dataset('iris')
iris.sample()
```

```python
titanic=pd.read_csv('train.csv')
titanic.sample()
```

### 1. Scatter plot (Num vs Num)

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

### 2. Barplot (num vs categ)

```python
sns.barplot(x=titanic['Pclass'],y=titanic['Age'])
```

### Boxplot (num vs categ)

```python
sns.boxplot(x=titanic['Sex'],y=titanic['Age'])
```

### 4. Distplot (num vs categ)

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'],hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'],hist=False)
```

### 5.Heatmap (categ vs categ)

```python
pd.crosstab(titanic['Pclass'],titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

### 6. Using groupby function (categ vs categ)

```python
titanic.groupby('Sex').mean(numeric_only=True)['Survived']
```

```python
sns.barplot(titanic.groupby('Sex').mean(numeric_only=True)['Survived'])
```

### 7. Clustermap (categ vs categ)

```python
sns.clustermap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

### 8. Pair plot

```python
sns.pairplot(iris)
```

### 8. lineplot (num vs num)

```python
new = flights.groupby('year').mean(numeric_only=True).reset_index()
new
```

```python
sns.lineplot(x=new['year'],y=new['passengers'])
```
